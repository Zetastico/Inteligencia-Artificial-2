# Dataset personalizado

PyTorch permite crear conjuntos de datos personalizados mediante la clase `Dataset`.

Se implementará una clase llamada `SkyviewDataset`, cuya función será:

- Leer las imágenes desde el disco.
- Convertirlas al formato RGB.
- Aplicar las transformaciones definidas durante el preprocesamiento.
- Asociar cada imagen con su etiqueta.
- Entregar los datos al `DataLoader` durante el entrenamiento.

Este enfoque proporciona un mayor control sobre el proceso de carga de datos y sigue la metodología presentada en el cuadernillo de entrenamiento.

```python
import torch
from torch.utils.data import Dataset
from torchvision import transforms
```

```python
class SkyviewDataset(Dataset):

    def __init__(self, image_paths, labels, transform=None):
        """
        Inicializa el dataset.

        Parámetros:
        - image_paths: lista con las rutas de las imágenes.
        - labels: lista con las etiquetas correspondientes.
        - transform: transformaciones que se aplicarán a cada imagen.
        """

        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        """
        Devuelve el número total de imágenes del dataset.
        """
        return len(self.image_paths)

    def __getitem__(self, idx):
        """
        Carga una imagen y su etiqueta correspondiente.
        """

        # Ruta de la imagen
        image_path = self.image_paths[idx]

        # Abrir imagen
        image = Image.open(image_path).convert("RGB")

        # Aplicar transformaciones
        if self.transform:
            image = self.transform(image)

        # Obtener la etiqueta
        label = self.labels[idx]

        return image, label
```

# División del Dataset

El conjunto de datos se dividirá en tres subconjuntos:

- **Entrenamiento (70 %):** utilizado para ajustar los pesos de la red neuronal.
- **Validación (15 %):** utilizado para monitorear el entrenamiento y aplicar técnicas como Early Stopping.
- **Prueba (15 %):** utilizado únicamente para evaluar el rendimiento final del modelo.

La división se realizará de forma **estratificada**, garantizando que todas las clases mantengan la misma proporción en cada subconjunto.

```python
# ==============================
# División Train (70%) y Temp (30%)
# ==============================

train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    image_paths,
    labels,
    test_size=0.30,
    random_state=SEED,
    stratify=labels
)
```

```python
# ==============================
# División Validation (15%) y Test (15%)
# ==============================

val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths,
    temp_labels,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_labels
)
```

# Transformaciones de las imágenes

Antes de ser utilizadas por la red neuronal, las imágenes deben pasar por una etapa de preprocesamiento.

Durante el entrenamiento se aplicarán las siguientes transformaciones:

- Redimensionar todas las imágenes a **224 × 224 píxeles**.
- Aplicar un volteo horizontal aleatorio como técnica de aumento de datos (*Data Augmentation*).
- Convertir las imágenes a tensores de PyTorch.
- Normalizar los valores de los píxeles para facilitar el proceso de aprendizaje.

Para los conjuntos de validación y prueba únicamente se aplicarán las transformaciones necesarias para garantizar una evaluación consistente, sin modificaciones aleatorias.

```python
# ==============================
# Transformaciones para entrenamiento
# ==============================

train_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])

# ==============================
# Transformaciones para validación y prueba
# ==============================

test_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])
```

```python
# Probar las transformaciones con una imagen

imagen = Image.open(image_paths[0]).convert("RGB")

imagen_transformada = train_transform(imagen)

print("Forma del tensor:", imagen_transformada.shape)
print("Tipo:", type(imagen_transformada))
```

# Arquitectura del modelo MLP

Para resolver el problema de clasificación se utilizará una red neuronal de tipo **Multilayer Perceptron (MLP)**.

La arquitectura propuesta está compuesta por:

- Una capa de entrada correspondiente a las imágenes de **224 × 224 × 3** píxeles.
- Tres capas ocultas totalmente conectadas.
- Función de activación **ReLU**.
- Capas **Batch Normalization** para estabilizar el entrenamiento.
- Capas **Dropout** para reducir el sobreajuste (*overfitting*).
- Una capa de salida con **15 neuronas**, una por cada clase del dataset.

Esta arquitectura busca un equilibrio entre capacidad de aprendizaje y costo computacional, siguiendo las buenas prácticas presentadas en el cuadernillo de entrenamiento.

# Técnicas de regularización y optimización

Para mejorar el entrenamiento de nuestra CNN y reducir el riesgo de sobreajuste,
utilizamos diferentes técnicas de regularización y optimización.

---

## 1. Técnicas de regularización

La **regularización** comprende técnicas que ayudan a evitar que el modelo
memorice demasiado los datos de entrenamiento y permiten mejorar su capacidad
de generalización sobre imágenes nuevas.

### 1.1 Dropout

Utilizamos la siguiente capa en el clasificador:

nn.Dropout(0.3)

Durante el entrenamiento, Dropout desactiva aleatoriamente aproximadamente el
30% de las activaciones.

Esto evita que el modelo dependa demasiado de determinadas neuronas y ayuda
a reducir el sobreajuste.

Sin Dropout:
Todas las neuronas participan

Con Dropout:
Algunas neuronas se desactivan aleatoriamente

El valor 0.3 indica la proporción de activaciones que se desactiva durante
el entrenamiento.

Objetivo: reducir el overfitting y mejorar la generalización.

### 1.2 Data Augmentation

También utilizamos una técnica de aumento de datos mediante:

transforms.RandomHorizontalFlip(p=0.5)

Esta transformación voltea horizontalmente una imagen con una probabilidad
del 50%.

De esta manera, el modelo puede observar diferentes variaciones de las
imágenes durante el entrenamiento.

Imagen original
      ↓
RandomHorizontalFlip
      ↓
Imagen original o volteada

Esto aumenta la variedad de los datos de entrenamiento y ayuda al modelo a
generalizar mejor.

Objetivo: aumentar la variedad de los datos y reducir el sobreajuste.

### 1.3 Early Stopping

Durante el entrenamiento también utilizamos Early Stopping.

Esta técnica detiene el entrenamiento cuando la pérdida de validación deja de
mejorar durante un número determinado de épocas.

En nuestro caso utilizamos un parámetro de paciencia:

patience=10

Esto significa que el entrenamiento puede detenerse después de varias épocas
sin una mejora en la validación.

Objetivo: evitar entrenar innecesariamente cuando el modelo deja de
mejorar sobre los datos de validación.

## 2. Técnicas de normalización

Batch Normalization

Nuestra CNN utiliza Batch Normalization después de cada convolución:

nn.BatchNorm2d(32)
nn.BatchNorm2d(64)
nn.BatchNorm2d(128)

Batch Normalization normaliza las activaciones de cada batch y ayuda a mantener
el entrenamiento más estable.

También puede contribuir a reducir ligeramente el sobreajuste, aunque su
función principal en nuestro modelo es estabilizar el entrenamiento.

Objetivo: estabilizar y facilitar el entrenamiento de la red.

## 3. Técnicas de optimización

Las técnicas de optimización se utilizan para modificar los pesos de la red
durante el entrenamiento con el objetivo de minimizar la función de pérdida.

### 3.1 Optimizador Adam

Utilizamos el optimizador:

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

Adam utiliza los gradientes calculados mediante backpropagation para actualizar
los pesos de la red.

De esta forma, después de calcular el error:

Predicción
    ↓
Loss
    ↓
Backpropagation
    ↓
Gradientes
    ↓
Adam
    ↓
Actualización de pesos

El parámetro lr corresponde al learning rate, que controla el tamaño de
los pasos realizados durante la actualización de los pesos.

Objetivo: actualizar los pesos del modelo para minimizar la función de
pérdida.

### 3.2 Learning Rate Scheduler

También utilizamos un scheduler para modificar el learning rate durante el
entrenamiento:

scheduler = torch.optim.lr_scheduler.StepLR(
    optimizer,
    step_size=15,
    gamma=0.5
)

StepLR reduce el learning rate después de un número determinado de épocas.

En nuestro caso:

step_size=15 → reduce el learning rate cada 15 épocas.
gamma=0.5 → multiplica el learning rate actual por 0.5.

Por ejemplo:

0.001
  ↓
0.0005
  ↓
0.00025
  ↓
...

Esto permite utilizar pasos más grandes al principio del entrenamiento y pasos
más pequeños posteriormente para realizar ajustes más precisos.

Objetivo: controlar el learning rate durante el entrenamiento y facilitar
la convergencia del modelo.

```python
import torch.nn as nn
```

```python
# ==============================
# Configuración del dispositivo
# ==============================

device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "mps" if torch.backends.mps.is_available()
    else "cpu"
)

print(f"Dispositivo utilizado: {device}")
```

```python
# ==============================
# Modelo MLP
# ==============================

class SkyviewMLP(nn.Module):

    def __init__(self, num_classes=15):
        super().__init__()

        self.model = nn.Sequential(

            # Aplanar la imagen
            nn.Flatten(),

            # Primera capa
            nn.Linear(224 * 224 * 3, 2048),
            nn.BatchNorm1d(2048),
            nn.ReLU(),
            nn.Dropout(0.3),

            # Primera capa
            nn.Linear(2048, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(),
            nn.Dropout(0.3),

            # Primera capa
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(),
            nn.Dropout(0.3),

            # Segunda capa
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.3),

            # Tercera capa
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.3),

            # Salida
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        return self.model(x)
```

```python


class SkyviewCNN(nn.Module):
    
    # Inicializa el modelo
    def __init__(self, num_classes=15):
        super().__init__()

        # Capas para extraer características
        self.features = nn.Sequential(

            # Bloque 1
            nn.Conv2d(3, 32, kernel_size=3, padding=1),  # Aplica 32 filtros
            nn.BatchNorm2d(32),                           # Normaliza los canales
            nn.ReLU(),                                   # Aplica activación ReLU
            nn.MaxPool2d(kernel_size=2, stride=2),       # Reduce el tamaño a la mitad

            # Bloque 2
            nn.Conv2d(32, 64, kernel_size=3, padding=1), # Aplica 64 filtros
            nn.BatchNorm2d(64),                           # Normaliza los canales
            nn.ReLU(),                                   # Aplica activación ReLU
            nn.MaxPool2d(kernel_size=2, stride=2),       # Reduce el tamaño a la mitad

            # Bloque 3
            nn.Conv2d(64, 128, kernel_size=3, padding=1), # Aplica 128 filtros
            nn.BatchNorm2d(128),                           # Normaliza los canales
            nn.ReLU(),                                    # Aplica activación ReLU
            nn.MaxPool2d(kernel_size=2, stride=2)          # Reduce el tamaño a la mitad
        )

        # Capas para clasificar las características
        self.classifier = nn.Sequential(

            nn.Flatten(),                    # Convierte los mapas en un vector

            nn.Linear(128 * 28 * 28, 256),  # Reduce las características a 256
            nn.ReLU(),                      # Aplica activación ReLU
            nn.Dropout(0.3),                # Desactiva el 30% de las neuronas

            nn.Linear(256, num_classes)     # Genera las 15 salidas
        )

    # Define el recorrido de los datos
    def forward(self, x):
        x = self.features(x)       # Extrae características
        x = self.classifier(x)     # Clasifica las características
        return x                   # Devuelve las predicciones
```

```python
# Crear el modelo

# model = SkyviewMLP(num_classes=len(classes)).to(device)

model = SkyviewCNN(num_classes=15).to(device)

print(model)
```

```python
single_dataset = torch.utils.data.Subset(
    train_dataset,
    [0]
)

single_loader = DataLoader(
    single_dataset,
    batch_size=1,
    shuffle=False
)
```

```python
single_model = SkyviewCNN(num_classes=15).to(device)

single_criterion = nn.CrossEntropyLoss()

single_optimizer = torch.optim.Adam(
    single_model.parameters(),
    lr=0.01
)
```

```python
# ==============================
# Fit de una sola imagen
# ==============================

epochs = 10

for epoch in range(1, epochs + 1):

    single_model.train()

    for x_batch, y_batch in single_loader:

        x_batch = x_batch.to(device)
        y_batch = y_batch.to(device)

        # Forward
        outputs = single_model(x_batch)

        # Loss
        loss = single_criterion(outputs, y_batch)

        # Backpropagation
        single_optimizer.zero_grad()
        loss.backward()
        single_optimizer.step()

        # Predicción
        prediction = torch.argmax(outputs, dim=1)

        print(
            f"Epoch {epoch}/{epochs} | "
            f"Loss: {loss.item():.5f} | "
            f"Real: {y_batch.item()} | "
            f"Pred: {prediction.item()}"
        )
```

```python
class SkyviewCNNDebug(nn.Module):

    def __init__(self, num_classes=15):
        super().__init__()

        self.features = nn.Sequential(

            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(128 * 28 * 28, 256),
            nn.ReLU(),

            # Sin Dropout para la prueba de sobreajuste

            nn.Linear(256, num_classes)
        )

    def forward(self, x):

        x = self.features(x)
        x = self.classifier(x)

        return x
```

```python
batch_model = SkyviewCNNDebug(num_classes=15).to(device)

batch_criterion = nn.CrossEntropyLoss()

batch_optimizer = torch.optim.Adam(
    batch_model.parameters(),
    lr=0.001
)
```

```python
# ==============================
# Fit de un batch
# ==============================

epochs = 200

for epoch in range(1, epochs + 1):

    batch_model.train()

    for x_batch, y_batch in batch_loader:

        x_batch = x_batch.to(device)
        y_batch = y_batch.to(device)

        # Forward
        outputs = batch_model(x_batch)

        # Loss
        loss = batch_criterion(outputs, y_batch)

        # Backpropagation
        batch_optimizer.zero_grad()
        loss.backward()
        batch_optimizer.step()

        # Accuracy del batch
        predictions = torch.argmax(outputs, dim=1)

        correct = (predictions == y_batch).sum().item()

        accuracy = correct / y_batch.size(0)

        if epoch % 10 == 0:

            print(
                f"Epoch {epoch}/{epochs} | "
                f"Loss: {loss.item():.5f} | "
                f"Accuracy: {correct}/{y_batch.size(0)} "
                f"({accuracy * 100:.2f}%)"
            )
```

# Entrenamiento del modelo

En esta sección se implementa la función encargada del entrenamiento de la red neuronal.

Durante cada época se realizarán las siguientes etapas:

1. Entrenamiento del modelo utilizando el conjunto de entrenamiento.
2. Evaluación sobre el conjunto de validación.
3. Registro de las métricas de entrenamiento y validación.
4. Actualización del Learning Rate mediante el Scheduler.
5. Aplicación de Early Stopping para evitar sobreajuste.
6. Almacenamiento del mejor modelo obtenido.

```python
history = {
    "train_loss": [],
    "train_acc": [],
    "val_loss": [],
    "val_acc": [],
    "learning_rate": []
}
```

```python
def train_model(
    model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    scheduler,
    epochs=50,
    patience=10
):

    best_val_loss = float("inf")
    epochs_without_improvement = 0

    history = {
        "train_loss": [],
        "train_acc": [],
        "val_loss": [],
        "val_acc": [],
        "learning_rate": []
    }

    for epoch in range(epochs):

        print(f"\nEpoch {epoch + 1}/{epochs}")

        # ------------------------
        # Entrenamiento
        # ------------------------

        model.train()

        running_loss = 0.0
        correct = 0
        total = 0

        for images, labels in train_loader:

            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            optimizer.step()

            running_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

        epoch_loss = running_loss / len(train_loader)
        epoch_acc = correct / total

        # ------------------------
        # Validación
        # ------------------------

        model.eval()

        val_loss = 0.0
        correct = 0
        total = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(device)
                labels = labels.to(device)

                outputs = model(images)

                loss = criterion(outputs, labels)

                val_loss += loss.item()

                _, predicted = torch.max(outputs, 1)

                total += labels.size(0)
                correct += (predicted == labels).sum().item()

        epoch_val_loss = val_loss / len(val_loader)
        epoch_val_acc = correct / total

        # ------------------------
        # Guardar mejor modelo
        # ------------------------

        if epoch_val_loss < best_val_loss:

            best_val_loss = epoch_val_loss

            torch.save(
                model.state_dict(),
                BEST_MODEL_PATH
            )

            print(
                f"✓ Mejor modelo guardado "
                f"(Val Loss = {best_val_loss:.4f})"
            )

            epochs_without_improvement = 0

        else:

            epochs_without_improvement += 1

        # ------------------------
        # Learning Rate
        # ------------------------

        current_lr = optimizer.param_groups[0]["lr"]

        # ------------------------
        # Guardar historial
        # ------------------------

        history["train_loss"].append(epoch_loss)
        history["train_acc"].append(epoch_acc)

        history["val_loss"].append(epoch_val_loss)
        history["val_acc"].append(epoch_val_acc)

        history["learning_rate"].append(current_lr)

        # ------------------------
        # Mostrar resultados
        # ------------------------

        print(
            f"Train Loss: {epoch_loss:.4f} | "
            f"Train Acc: {epoch_acc:.4f} | "
            f"Val Loss: {epoch_val_loss:.4f} | "
            f"Val Acc: {epoch_val_acc:.4f} | "
            f"LR: {current_lr:.6f}"
        )

        # ------------------------
        # Scheduler
        # ------------------------

        scheduler.step()

        # ------------------------
        # Early Stopping
        # ------------------------

        if epochs_without_improvement >= patience:

            print("\nEarly Stopping activado.")

            break

    # ------------------------
    # Cargar mejor modelo
    # ------------------------

    model.load_state_dict(
        torch.load(
            BEST_MODEL_PATH,
            map_location=device
        )
    )

    model.eval()

    return model, history
```

Se recorren todas las imágenes del conjunto de prueba para obtener:

Estas variables serán utilizadas posteriormente para calcular todas las métricas de clasificación.

```python
def get_predictions(model, test_loader, device):

    model.eval()

    y_true = []
    y_pred = []
    y_prob = []

    with torch.no_grad():

        for images, labels in test_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            probabilities = torch.softmax(outputs, dim=1)

            _, predicted = torch.max(probabilities, dim=1)

            y_true.extend(labels.cpu().numpy())

            y_pred.extend(predicted.cpu().numpy())

            y_prob.extend(probabilities.cpu().numpy())

    return (
        np.array(y_true),
        np.array(y_pred),
        np.array(y_prob)
    )
```
